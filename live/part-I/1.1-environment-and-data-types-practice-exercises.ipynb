## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm). Print each result to two decimals with an f-string.

````python
# Your solution here
temperature_celsius = 18.5
temperature_kelvin = temperature_celsius + 273.15
temperature_fahrenheit = temperature_celsius * 9 / 5 + 32
print(f"Temperature in Kelvin: {temperature_kelvin:.2f} K")
print(f"Temperature in Fahrenheit: {temperature_fahrenheit:.2f} °F")
discharge_m3_per_s = 4.7
area_km2 = 250
specific_discharge_mm_per_day = discharge_m3_per_s * 1000 / area_km2 / 864
print(f"Specific discharge in mm/day: {specific_discharge_mm_per_day:.2f} mm/day")
````
